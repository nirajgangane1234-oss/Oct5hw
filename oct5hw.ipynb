```python
# WAP to find a name starts with specific character "A"

names = ["Niraj","Arpan","Abhinay","Aayush","Sahil","Vedant","Hrishi","Shantanu"]
for name in names:
    if name.startswith("A"):
        print(name)
```

```python
# WAP to Check names  having character more than 5

names = ["Niraj","Arpan","Abhinay","Aayush","Sahil","Vedant","Hrishi","Shantanu"]
for name in names:
    if len(name) > 5:
        print(name)
```

```python
# WAP to count a specific character like "a"

names = ["Niraj","Arpan","Abhinay","Aayush","Sahil","Vedant","Hrishi","Shantanu"]
count = 0

for name in names:
    count += name.lower().count("a")

print(count)
```
